```python
import pandas as pd
import glob
import os

# 获取你要处理的文件路径和时间列名
# 假设你想处理 ds4/0.05/ 下的文件，可以把路径替换成 ds4_in_path
folder_path = ds4_in_path 
time_col = ds4_parameters['time_name']

print(f"🧹 开始清洗 {folder_path} 下的数据...")

# 找到该目录下所有的 csv 文件
csv_files = glob.glob(os.path.join(folder_path, '*.csv'))

for file in csv_files:
    # 1. 读取 CSV
    df = pd.read_csv(file)
    
    # 2. 强制转换时间列，使用 format='mixed' 兼容有毫秒和没毫秒的情况
    # 然后统一格式化为 "%Y-%m-%d %H:%M:%S.%f"
    df[time_col] = pd.to_datetime(df[time_col], format='mixed').dt.strftime('%Y-%m-%d %H:%M:%S.%f')
    
    # 3. 覆盖保存回原文件
    df.to_csv(file, index=False)
    print(f"✅ 已清洗并统一时间格式: {file}")

print("🎉 所有文件时间格式清洗完毕！现在可以安全运行 trainfiles_generation 了！")
```
