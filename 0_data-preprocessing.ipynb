# DS3


```python
ds3 = pd.read_csv('OCEL/angular_github_commits_ocel.csv', sep=',')
with pd.option_context('display.max_columns', None): 
    display(ds3)
ds3['ocel:timestamp'] = pd.to_datetime(ds3['ocel:timestamp'], errors='coerce')
ds3 = ds3.dropna(subset=['ocel:timestamp'])
```

```python
ds3 = ds3.drop(columns=["Unnamed: 10","Unnamed: 0","Unnamed: 11","commit_message"])
ds3.columns = ds3.columns.str.replace('ocel:', '')
ds3.columns = ds3.columns.str.replace('type:', '')
ds3 = ds3.rename(columns={'timestamp': 'event_timestamp', 'activity': 'event_activity', "eid": "event_id"})
ds3['event_id'] = range(len(ds3))
display(ds3)
```

```python
objects = ['branches', 'files']
first_cols = ['event_id', 'event_timestamp', 'event_activity'] + objects
split = []
ds3 = preprocess_dataframe(ds3, first_cols, split)
ds3['event_timestamp'] = pd.to_datetime(ds3['event_timestamp'], errors='coerce', utc=True)
ds3['event_timestamp'] = ds3['event_timestamp'].dt.tz_localize(None)
ds3['event_timestamp'] = ds3['event_timestamp'].dt.strftime('%Y/%m/%d %H:%M:%S')
display(ds3)
```

```python
directory = 'data-prepro'
filename = directory + '/ds3_prepro.csv'

if not os.path.exists(directory):
    os.makedirs(directory)

ds3.to_csv(filename, index=False)
```
